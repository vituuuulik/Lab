```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np 
import statsmodels.api as sm
df = pd.read_excel('join.xlsx')
df = df.dropna(subset = ['Генотип rs6295']) #убираем пустые значения в колонке генотип
df['Генотип rs6295'] = df['Генотип rs6295'].map({'G/G' : 0, 'G/C' : 1, 'C/C' : 2}) #изменяем значения в каждом столбце
#переводим формат даты в формат для питона
df['2. Дата рождения'] = pd.to_datetime(df['2. Дата рождения'])
df['5. Дата приема участника исследования'] = pd.to_datetime(df['5. Дата приема участника исследования'])
#вычисляем возраст на момент поступления (dt.year - это доступ к году из даты)
df['Возраст на момент поступления'] = df['5. Дата приема участника исследования'].dt.year - df['2. Дата рождения'].dt.year

#поправляем погрешность, true.astype(int) = 1, а False = 0; если True, то вычтется 1 год
df['Возраст на момент поступления'] -= ((df['5. Дата приема участника исследования'].dt.month < df['2. Дата рождения'].dt.month) | 
                                        ((df['5. Дата приема участника исследования'].dt.month == df['2. Дата рождения'].dt.month) & 
                                         (df['5. Дата приема участника исследования'].dt.day < df['2. Дата рождения'].dt.day))
                                         ).astype(int)

sns.boxplot(data = df, x = 'Генотип rs6295', y = 'Возраст на момент поступления')
plt.xlabel('Генотип')
plt.ylabel('Возраст')
plt.title('Распределение SNP по возрасту')
plt.show()

df_filtered = df[df['Выберите группу'].isin(['Группа гетероагрессия', 'Группа контроля'])].copy()
df_filtered['Выберите группу'] = df_filtered['Выберите группу'].map({'Группа контроля' : 0, 'Группа гетероагрессия' : 1})

sns.countplot(data=df_filtered, x= 'Генотип rs6295', order = [0, 1, 2])
plt.xlabel('Генотип')
plt.ylabel('количество пациентов гетероагрессией')
plt.title('Зависимость гетероагрессии от генотипа')
plt.show()

#считаем регрессию
df_for_reg = df_filtered[['Выберите группу', 'Генотип rs6295' ]]

x = df_for_reg[['Генотип rs6295']] #в [[]] тк add_constant ждёт именно DataFrame, а не Series 
x = sm.add_constant(x) #добавляем независимую переменную
y = df_for_reg['Выберите группу'] 
model = sm.Logit(y, x).fit() #fit() - ищет лучшие коэф b0 и b1
params = model.params #b0 и b1 
conf = model.conf_int() #доверительный интервалл

#экспоненцируем 
or_value = np.exp(params) #OR
or_ci_lower = np.exp(conf[0])
or_ci_upper = np.exp(conf[1])
p_value = model.pvalues['Генотип rs6295']

print(f"OR = {or_value['Генотип rs6295']:.3f}")
print(f"Доверительный интервалл:{or_ci_lower['Генотип rs6295']:.3f} - {or_ci_upper['Генотип rs6295']:.3f}")
print(f"p_value = {p_value:.3f}")
```
