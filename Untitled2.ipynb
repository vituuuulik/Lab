```python
import pandas as pd
import numpy as np
from scipy.stats import chi2
from scipy.stats import chisquare
df1 = pd.read_excel('big.xlsx')
df2 = pd.read_excel('small.xlsx')
df_join = pd.merge(df1, df2, left_on = '4. Идентификационный номер участника исследования', right_on = 'ПАЦИЕНТ', how = 'left')
df_join = df_join.drop('ПАЦИЕНТ', axis = 1)
df_join_memory = df_join.copy()

#переделать на 
count_deleted = int(input('сколько групп вы хотите исключить из таблицы:'))
if count_deleted != 0:
    for i in range(count_deleted):
        delete_axis = input('введите название столбца с группой исключения:')
        delete_group = input('введите название группы исключения:')
        print(f'удаляем строки где {delete_axis} = {delete_group} ')
        df_join = df_join[df_join[delete_axis] != delete_group]
```

```python
#известные частоты аллелей в популяции
G_pop = float(input('введите распространённость аллелей G в популяции в долях (округлите до десятых):'))  
C_pop = float(input('введите распространённость аллелей C в популяции в долях (округлите до десятых):'))
if G_pop + C_pop != 1: #пока подумаем
    print('ошибка, повторите ввод')
    G_pop = float(input('введите распространённость аллелей G в популяции:'))  
    C_pop = float(input('введите распространённость аллелей C в популяции:'))
    

```

```python
#наблюдаемое количество генотипов GG, GC и СС 
n_GG = df_join['Генотип rs6295'].value_counts()['G/G']
n_GC = df_join['Генотип rs6295'].value_counts()['G/C']
n_CC = df_join['Генотип rs6295'].value_counts()['C/C']
N = n_GG + n_GC + n_CC
observed = [n_GG, n_GC, n_CC]
#ожидаемое количество генотипов по частотам аллелей из популяции
exp_GG = N*(G_pop)**2
exp_GC = N*(2*(G_pop)*(C_pop))
exp_CC = N*(C_pop)**2

expected = [exp_GG, exp_GC, exp_CC]
```

```python
#считаем chi2 и p-value
#ddof - степени свободы 
chi2_stat, p_value = chisquare(observed, expected, ddof = 1 ) 
chi_critical = chi2.ppf(0.9, df = 1)
print(chi2_stat)
print(chi_critical)
```

```python
if chi2_stat < chi_critical: #на 15% больше chi_critical (погрешность)
    print('соответствует раcпределению Харди-Вайнберга')
    print(f'количество гомозигот по аллелю G: {n_GG}')
    print(f'количество гетерозигот (G/C): {n_GC}')
    print(f'количество гомозигот по аллелю C: {n_CC}')
    print(f'из таблицы удалено {len(df_join_memory)-len(df_join)} пациентов')
else:
    print('не соответствует раcпределению Харди-Вайнберга')
    print(f'количество гомозигот по аллелю G: {n_GG}')
    print(f'количество гетерозигот (G/C): {n_GC}')
    print(f'количество гомозигот по аллелю C: {n_CC}')
    print(f'из таблицы удалено {len(df_join_memory)-len(df_join)} пациентов')
```
